```python
import pandas as pd

# 1. Membaca file CSV
df = pd.read_csv('paket_a_transaksi_pos_2024_2025.csv')

# 2. Menangani Missing Value pada kolom penting
df['jumlah'] = df['jumlah'].fillna(0)
df['harga_satuan'] = df['harga_satuan'].fillna(0)
df['diskon_persen'] = df['diskon_persen'].fillna(0)
df['status'] = df['status'].fillna('UNKNOWN')

# 3. FIX UTAMA: Ubah 'harga_satuan' menjadi angka (float), bersihkan tanda koma jika ada
df['harga_satuan'] = df['harga_satuan'].astype(str).str.replace(',', '.').astype(float)

# 4. Hati-hati dengan STATUS: Filter hanya status yang 'SELESAI'
df_valid = df[df['status'] == 'SELESAI'].copy()

# 5. Hati-hati dengan DISKON: Hitung total pendapatan per baris
df_valid['pendapatan'] = df_valid['jumlah'] * df_valid['harga_satuan'] * (1 - df_valid['diskon_persen'] / 100)

# 6. Mengubah kolom tanggal dan mengambil tahunnya
df_valid['tanggal'] = pd.to_datetime(df_valid['tanggal'], format='mixed', dayfirst=True)
df_valid['tahun'] = df_valid['tanggal'].dt.year

# 7. Menghitung jumlah pendapatan per tahun (2024 dan 2025) dipisah
pendapatan_2024 = df_valid[df_valid['tahun'] == 2024]['pendapatan'].sum()
pendapatan_2025 = df_valid[df_valid['tahun'] == 2025]['pendapatan'].sum()

# Menampilkan nilai saja sesuai permintaan tugas
print(f"Pendapatan 2024: {pendapatan_2024}")
print(f"Pendapatan 2025: {pendapatan_2025}")
```
